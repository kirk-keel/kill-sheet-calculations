# Case Study 05: Tubular Buckling Limits & Bit Hydraulics Optimization

## 1. Operational Background
In high-angle extended reach drilling (ERD) and horizontal lateral stimulation/milling:
1. **Buckling Stability:** Compressive forces on the drill string or coiled tubing induce sinusoidal waving (Dawson-Paslay criteria) followed by helical lockup (Wu & Juvkam-Wold / Lubinski criteria), where wall friction prevents further weight transfer to the bit.
2. **Bit Hydraulics:** Bit nozzle sizes dictate Total Flow Area (TFA), bottomhole cleaning Jet Impact Force (JIF), and hydraulic horsepower per square inch (HSI), balanced against annular friction pressure loss and Equivalent Circulating Density (ECD).

This case study demonstrates:
- Critical compressive load evaluation across varying wellbore inclinations
- Nozzle jet sizing and power optimization
- Annular friction loss and circulating ECD calculation

```python
from killsheet.buckling import sinusoidal_buckling_load_lbf, helical_buckling_load_lbf, evaluate_tubular_buckling
from killsheet.hydraulics import bit_hydraulics_report, annular_pressure_loss_psi, equivalent_circulating_density_ppg

# Drillstring & Wellbore Parameters
PIPE_OD = 5.0
PIPE_ID = 4.276
HOLE_ID = 8.5
PIPE_WEIGHT = 19.5
MW_PPG = 10.5

print("Inclination (deg) | Sinusoidal Limit (lbf) | Helical Limit (lbf)")
print("-" * 60)
for inc in [15, 30, 45, 60, 75, 90]:
    f_sin = sinusoidal_buckling_load_lbf(PIPE_OD, PIPE_ID, HOLE_ID, PIPE_WEIGHT, MW_PPG, inc)
    f_hel = helical_buckling_load_lbf(PIPE_OD, PIPE_ID, HOLE_ID, PIPE_WEIGHT, MW_PPG, inc)
    print(f"{inc:>17} | {f_sin:>22,} | {f_hel:>19,}")
```

## 2. Bit Hydraulics & Nozzle Jet Optimization

```python
flow_rate_gpm = 450.0
bit_diameter_in = 8.75
nozzle_combos = {
    "3 x 11/32\"": [11, 11, 11],
    "3 x 12/32\"": [12, 12, 12],
    "3 x 13/32\"": [13, 13, 13],
}

print("Nozzle Array | TFA (sq in) | Bit dP (psi) | HHP (hp) | HSI (hp/sq in) | JIF (lbf)")
print("-" * 78)
for name, nozzles in nozzle_combos.items():
    rep = bit_hydraulics_report(MW_PPG, flow_rate_gpm, bit_diameter_in, nozzles)
    print(f"{name:<12} | {rep['tfa_sq_in']:>11.4f} | {rep['bit_pressure_drop_psi']:>12,} | {rep['hydraulic_horsepower_hhp']:>8.1f} | {rep['hsi_hp_per_sq_in']:>14.2f} | {rep['jet_impact_force_lbf']:>9,}")
```
