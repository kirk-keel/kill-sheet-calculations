# Case Study 04: Casing Seat Selection & Kick Tolerance Analysis

## 1. Operational Background
During exploration and development drilling, the casing shoe depth must be chosen such that the formation fracture resistance at the shoe can withstand both:
1. The hydrostatic pressure of the drilling mud.
2. The maximum trapped annular pressure when circulating out an influx under constant bottomhole pressure (Wait & Weight or Driller's Method).

Under IADC WellSharp standards and Boyle's Real Gas Law ($P_1 V_1 = P_2 V_2$), the critical risk condition occurs when the top of the gas kick bubble reaches the casing shoe.

This case study demonstrates:
- Determination of kick intensity from shut-in drill pipe pressure (SIDPP)
- Calculation of maximum allowable annular surface pressure (MAASP)
- Determination of maximum permissible gas influx height at the shoe ($H_{\text{max}}$)
- Calculation of bottomhole kick tolerance volume ($V_{\text{kick}}$)

```python
from killsheet.kick_tolerance import kick_tolerance_summary, kick_intensity_ppg, max_gas_height_at_shoe_ft, kick_tolerance_bottomhole_bbl

# Well Parameters
MUD_WEIGHT_PPG = 12.5
BIT_TVD_FT = 10_000
SHOE_TVD_FT = 5_000
LOT_EMW_PPG = 14.5
SIDPP_PSI = 350
ANNULAR_CAP_BBL_PER_FT = 0.0459  # 8-1/2" hole with 5" drill pipe

kt_report = kick_tolerance_summary(
    mud_weight_ppg=MUD_WEIGHT_PPG,
    tvd_ft=BIT_TVD_FT,
    shoe_tvd_ft=SHOE_TVD_FT,
    lot_emw_ppg=LOT_EMW_PPG,
    sidpp_psi=SIDPP_PSI,
    annular_capacity_bbl_per_ft=ANNULAR_CAP_BBL_PER_FT
)

for key, val in kt_report.items():
    print(f"{key:<35}: {val}")
```

## 2. Sensitivity Analysis: Kick Tolerance vs. Casing Shoe Setting Depth
As drilling progresses deeper below the casing shoe, the available kick tolerance decreases due to higher pore pressures and longer open-hole intervals.

```python
shoe_depths = [4000, 5000, 6000, 7000]
print("Shoe TVD (ft) | MAASP (psi) | Max Gas Ht (ft) | Kick Tolerance (bbl)")
print("-" * 60)
for shoe in shoe_depths:
    res = kick_tolerance_summary(
        mud_weight_ppg=MUD_WEIGHT_PPG,
        tvd_ft=BIT_TVD_FT,
        shoe_tvd_ft=shoe,
        lot_emw_ppg=LOT_EMW_PPG,
        sidpp_psi=SIDPP_PSI,
        annular_capacity_bbl_per_ft=ANNULAR_CAP_BBL_PER_FT
    )
    print(f"{shoe:>13} | {res['shoe_maasp_psi']:>11} | {res['max_gas_height_at_shoe_ft']:>15} | {res['kick_tolerance_bottomhole_bbl']:>20.1f}")
```
